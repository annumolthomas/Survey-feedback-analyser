STEP 3:Text Cleaning

```python
#Clean all feedback entries 
def clean_feedback_data(d):
#removing punctuations
    d = d.replace('.', '')
    d = d.replace(',' ,'')
    d = d.replace('!', '')
    d = d.replace('?', '')

#replace multiple space with single space:
    d = ' '.join(d.split())
#convert to lower case
    d = d.lower()

    return d
#we have to clean all feedback entries=len

for i in range(len(feedback_data['Feedback'])):
    feedback_data['Feedback'][i] = clean_feedback_data(
        feedback_data['Feedback'][i])
result = clean_feedback_data(feedback_data['Feedback'][i]) 

print(feedback_data['Feedback'])
```

STEP 4: Count Words

```python
#Create a function count_word_in_feedbacks(word) that:  Takes a word as input. 


def count_word_in_feedbacks(word):
    count = 0
    target = word.lower()
    
    for feedback in feedback_data['Feedback']:
        feedback_words =feedback.lower().split()#check case sensitive and exact match 
        if target in feedback_words:
            count += 1
    return count

print("Word Counts")

#Number of feedbacks containing "good"

print("Good: ",
      count_word_in_feedbacks("good"))

#Number of feedbacks containing "poor"

print("Poor: ",
      count_word_in_feedbacks("poor"))

#Number of feedbacks containing "excellent"

print("Excellent: ",
      count_word_in_feedbacks("excellent"))
```
